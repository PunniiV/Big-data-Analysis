## Big Data Analysis using PySpark - Netflix Dataset

This notebook uses PySpark to perform analysis on the Netflix dataset.

**Dataset:** `netflix_titles.csv`

**Objectives:**
- Load and inspect the dataset
- Clean and preprocess data
- Analyze top genres, release years, and countries
- Derive meaningful insights from the dataset

```python
# Install PySpark if needed
!pip install pyspark
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, count, year, desc, split, explode

# Start Spark session
spark = SparkSession.builder.appName("NetflixBigDataAnalysis").getOrCreate()
```

```python
# Load the Netflix dataset
df = spark.read.csv("/mnt/data/netflix_dataset/netflix_titles.csv", header=True, inferSchema=True)

# Print schema
df.printSchema()
```

```python
# Show sample records
df.show(5)
```

```python
# Count total records and nulls in each column
df.count(), [(col_name, df.filter(col(col_name).isNull()).count()) for col_name in df.columns]
```

```python
# Top 10 content-producing countries
df.groupBy("country").count().orderBy(desc("count")).show(10)
```

```python
# Top 10 most common genres (listed under 'listed_in')
df_genres = df.withColumn("genre", explode(split(col("listed_in"), ", ")))
df_genres.groupBy("genre").count().orderBy(desc("count")).show(10)
```

```python
# Number of titles released by year
df_year = df.withColumn("release_year", year("date_added"))
df_year.groupBy("release_year").count().orderBy("release_year").show()
```

```python
# Count of Movies vs TV Shows
df.groupBy("type").count().show()
```

```python
# Stop Spark session
spark.stop()
```

## 📌 Insights Summary

- **Most common genre**: Listed under 'listed_in' after exploding the values.
- **Top producing country**: Based on the 'country' column.
- **Content trend over time**: Number of titles added per year.
- **Type distribution**: Number of movies vs TV shows.
